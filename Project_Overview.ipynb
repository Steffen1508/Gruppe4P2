# Project Overview: PII Detection Pipeline (Group 4, P2 Project)

## 1. Project Goal & Context
The goal of this project is to develop a robust, high-performance Machine Learning pipeline for **Personally Identifiable Information (PII) detection** in text documents (e.g., PDFs). This project evaluates multiple approaches ranging from simple regular expressions to classical ML (SVM with TF-IDF) and state-of-the-art Deep Learning models (BERT Token Classification).

**Key Non-Functional Requirements (NFRs) Met:**
- **Latency**: Inference must be under 100ms per request. The current BERT v6 achieves ~6-10ms latency on GPU and ~23ms on average for CPU fallbacks.
- **Accuracy**: Targeting high F1 scores across critical PII categories (Financial, Identity, Personal Info).

## 2. Current Progress & Achievements
The project has successfully progressed through several iterations:
- **Data Pipeline**: A robust `data_loader.py` securely combines `syvai/pii-dataset-eng` and `nvidia/Nemotron-PII`, utilizing structured data filtering and credit card formatting augmentations.
- **BERT Model Evolution (v1 to v6)**:
  - Transitioned from sequence-level to token-level classification.
  - Resolved class imbalance via class weighting, label reduction (from 25 to 12 core priority labels), and data augmentation.
  - Final metrics for **BERT v6**: Macro F1: **0.94**, Weighted F1: **1.00**.
- **Classical ML Baseline**: Functional TF-IDF and `SGDClassifier` (SVM) training loop implemented in `TF-IDF/tfidf_model.py` handling multi-label classification.
- **Inference Pipeline**: `BERT_inference.py` loads the trained weights and extracts entities straight from raw PDFs using `pypdf`.

## 3. System Architecture & Modules
Currently, the codebase is separated into experimental folders, but moving towards a unified architecture.

* **`data_loader.py`**: The central data ingestion module. Combines HF datasets, normalizes text, and masks/unmasks tabular formats.
* **`pipeline.py` (WIP)**: Intended to be the standard wrapper unifying Regex, SVM, and BERT predictions under a single interface `Pipeline`.
* **`BERT/` Directory**: Contains the isolated research loop for the deep learning component, including tests against external datasets (`BERT_nemotron_test.py`) and PyTorch training scripts with AMP (Automatic Mixed Precision).
* **`TF-IDF/` Directory**: Contains classical baseline implementations for performance and latency benchmarking.
* **`Tokenization_og_Chunking/`**: Explores how text is optimally split before entering the vectorizers or Transformer models.

## 4. Identified Gaps & Next Steps
As an ML architect reviewing this repository, the technical rigor of the model training is strong. However, to bring this from "research scripts" to a "finished master's project", the following software engineering and structural gaps must be addressed:

### Gap 1: Unification of the Evaluation Strategy
Currently, `pipeline.py` is a placeholder for BERT inference, and models are evaluated in separate scripts (`TF-IDF/tfidf_model.py` vs `BERT/BERT_nemotron_test.py`). 
**Action**: Build a unified evaluation framework that passes the exact same test set (e.g., an unseen slice of Nemotron) to Regex, SVM, and BERT, and generates a side-by-side comparative classification report.

### Gap 2: Finalizing `pipeline.py`
The overarching `Pipeline` class has dummy implementations for BERT loading (`"bert_model_pladceholder"`).
**Action**: Integrate the Hugging Face `pipeline` or custom `BERT/BERT_inference.py` logic directly into `pipeline.py` so the whole project can be run from one entry point.

### Gap 3: Code Duplication & Consistency
We have `data_loader.py` at the root and `BERT/data_loader.py`. 
**Action**: Consolidate data loading to a single source of truth in the root directory to ensure all models are trained on the exact same data distributions.
